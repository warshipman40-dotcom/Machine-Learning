```python
from sklearn.datasets import load_iris
iris = load_iris()
```

```python
#flower measurements (input data) of this iris flowers
#each one is a list
X = iris.data
#target label
#answers are 0, 1, 2 since there are 3 types of iris
y = iris.target
feature_names = iris.feature_names
target_names = iris.target_names
print(feature_names)
print(target_names)
```

```python
from sklearn.model_selection import train_test_split
#we split the data because the model needs to learn from some data but also test on some data it hasn't seen
#splits data into 4 parts (test_size = 0.4, (40% of data goes to test set, 60% goes to training set)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2)
#X_train (flower measurements for learning, y_train (correct answers)
#X_test (flower measurements for testing), y_test (correct answers for testing)
#.shape gives dimensions (rows, features) of data
#example (90, 4) means 90 rows, 4 features per flower
#example X_Train returns (60, 4) means we have 60 rows of training data
print(X_train.shape)
#X_test returning (90, 4) means we have 90 rows of input to test
print(X_test.shape)
```

```python
from sklearn.neighbors import KNeighborsClassifier
#creates the model (calculates the closest neighbors)
#for example measurements are [5.0, 1.5, 3.2, 1.5] and three closest flowers are Setosa, Setosa, Versicolor
#using majority voting the model predicts Setosa
#we use 3 because there are only 3 types of iris flower
knc = KNeighborsClassifier(n_neighbors = 3)
#.fit() trains the model
#stores examples of flowers data and their correct data to compare with new flowers
knc.fit(X_train, y_train)
#model looks at the remaining flowers in X_test, finds the 3 closest training flowers, and votes on the flower type
#predictions of flower types are storted in y_pred
y_pred = knc.predict(X_test)
```

```python
from sklearn import metrics
#compares actual answers with predicted answers and gives a % accuracy score
print(metrics.accuracy_score(y_test, y_pred))
```

```python
sample = [[3, 5, 4, 2], [2, 3, 5, 4]]
predictions = knc.predict(sample)
pred_species = [iris.target_names[p] for p in predictions]
print("predictions : ", pred_species)
```

```python
from sklearn.externals import joblib
#stores the trained model in a binary file
joblib.dump(knc, "mlbrain.joblib")
#to get our model we can do joblib.load("mlbrain.joblib")
```
